# Data Tupe of pandas : Series , DataFrame

### Series : 1 Dimensional Column of data

```python
Car = pd.Series(["Toyota" , "BMW" , "Mercedes"])
Car
```

```python
Color = pd.Series(["Red" , "Blue" , "Black"])
Color
```

# Data Frame : Row and Column

```python
Car_Data = pd.DataFrame({
    "Car_Type" : Car , 
    "Car_Color" : Color
})
Car_Data
```

# Practice Exercise

```python
Foods = pd.Series(["Pizza" , "Burger" , "Shawrma" , "Rise"])
Prices = pd.Series([10 , 20 , 5 , 6])
Foods_Menu = pd.DataFrame(
    {
        "Food_Types" : Foods,
        "Price_in_Dollar" : Prices
    }
)
Foods_Menu
```

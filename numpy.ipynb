# Anatomy of array 

Array : A list of number can be multiple dimension
Scalar : a single number 
Vector : a list of number with 1 dimenstion (1-d array)
Matrix : a list  of multiple dimension (n-d array)

# Attribute of an array

.shape : the exact size along each dimension (Rows, Column)
.ndim : the total number of dimensions
.dtype : the type of data stored inside 
.size : the total count of all elements inside
